## 1. 基础代码执行示例

这个示例演示了 JupyterHub 中的基本代码执行和数据可视化。

```python
import numpy as np
import matplotlib.pyplot as plt

# 创建数据
x = np.linspace(0, 10, 100)
y = np.sin(x)

# 绘图
plt.figure(figsize=(5, 3))
plt.plot(x, y, label='sin(x)')
plt.title('Sine Function Plot')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 2. 交互式可视化示例

这个示例演示了如何使用 ipywidgets 创建交互式可视化。

```python
from ipywidgets import interact
import numpy as np


def plot_sine(frequency=1.0, amplitude=1.0):
    x = np.linspace(0, 2*np.pi, 201)
    y = amplitude * np.sin(frequency * x)
    plt.figure(figsize=(5, 3))
    plt.plot(x, y, label='sin(x)')
    plt.ylim(-5, 5)
    plt.title(f'Sine Function (f={frequency}, A={amplitude})')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend(loc='lower left')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()


# 创建交互式控件
interact(plot_sine, frequency=(1, 10, 0.5), amplitude=(0.25, 5, 0.25))
```

## 3. 高级数据处理示例

演示更复杂的数据处理和可视化操作。

```python
import pandas as pd

# 创建示例数据
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [120, 150, 180, 220, 190, 250],
    'Expenses': [80, 90, 100, 110, 95, 120]
}

df = pd.DataFrame(data)
print("数据表:")
print(df)
print()
print(f"总销售额: {df['Sales'].sum()}")
print(f"平均销售额: {df['Sales'].mean():.2f}")
```

```python
# 绘制数据图表
plt.figure(figsize=(6, 5))
plt.plot(df['Month'], df['Sales'], marker='o', linewidth=2, label='Sales')
plt.plot(df['Month'], df['Expenses'], marker='s', linewidth=2, label='Expenses')
plt.title('Monthly Sales and Expenses', fontsize=14)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Amount', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 4. Notebook 使用技巧

### 单元格管理
- 按 `A` 在上方插入单元格（命令模式下）
- 按 `B` 在下方插入单元格（命令模式下）
- 按 `DD` 删除当前单元格（命令模式下）
- 按 `M` 将单元格转换为 Markdown 类型（命令模式下）
- 按 `Y` 将单元格转换为 Code 类型（命令模式下）

### 代码执行
- `Shift+Enter`: 运行单元格并移动到下一个
- `Ctrl+Enter`: 运行单元格并保持当前位置
- `Alt+Enter`: 运行单元格并在下方插入新单元格

### 模式切换
- `Esc`: 进入命令模式
- `Enter`: 进入编辑模式

### 获取帮助
- 在函数名后加 `?` 显示文档
- `Shift+Tab` 显示快速帮助
- `Tab` 显示代码补全选项
