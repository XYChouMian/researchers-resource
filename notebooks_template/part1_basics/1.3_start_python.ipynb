### 修改列表

````python
# 修改和操作列表
fruits = ["苹果", "香蕉", "橙子"]
print(f"原始: {fruits}")

# 修改
fruits[1] = "草莓"
print(f"修改后: {fruits}")

# 添加
fruits.append("葡萄")
print(f"添加后: {fruits}")

# 删除
fruits.remove("苹果")
print(f"删除后: {fruits}")

# 弹出
last = fruits.pop()
print(f"弹出: {last}")
print(f"最终: {fruits}")
````

````python
# 遍历列表
fruits = ["苹果", "香蕉", "橙子"]

print("for循环遍历:")
for fruit in fruits:
    print(f"- {fruit}")

print("\n索引遍历:")
for i in range(len(fruits)):
    print(f"索引{i}: {fruits[i]}")

print("\nenumerate遍历:")
for index, fruit in enumerate(fruits):
    print(f"索引{index}: {fruit}")
````

## 6. 条件判断
### if语句

````python
# 基本if语句
score = 85

if score >= 90:
    print("优秀")
elif score >= 80:
    print("良好")
elif score >= 60:
    print("及格")
else:
    print("不及格")
````

## 7. 循环
### for循环

````python
# for循环遍历
fruits = ["苹果", "香蕉", "橙子"]

print("遍历列表:")
for fruit in fruits:
    print(f"- {fruit}")

print("\n遍历数字:")
for i in range(1, 6):
    print(i)

print("\n遍历字符串:")
text = "Python"
for char in text:
    print(f"- {char}")
````

### while循环

````python
# while循环
count = 5
print("倒计时:")
while count > 0:
    print(count)
    count -= 1
print("发射！")

# 求和
total = 0
i = 1
while i <= 5:
    total += i
    i += 1
print(f"\n1+2+3+4+5 = {total}")
````

````python
# 循环控制
print("break示例:")
for i in range(1, 10):
    if i == 5:
        break
    print(i)

print("\ncontinue示例:")
for i in range(1, 10):
    if i == 5:
        continue
    print(i)

print("\n循环中的else:")
for i in range(1, 5):
    print(i)
else:
    print("循环正常完成")
````

## 8. 函数
### 定义和调用

````python
# 简单函数
def greet(name):
    print(f"你好，{name}！")

greet("小明")
greet("张三")
````

````python
# 带返回值的函数
def calculate_rectangle(length, width):
    area = length * width
    perimeter = 2 * (length + width)
    return area, perimeter

a, p = calculate_rectangle(5, 3)
print(f"面积: {a}")
print(f"周长: {p}")

# 也可以作为元组接收
result = calculate_rectangle(4, 2)
print(f"结果: {result}")
````

````python
# 带默认参数的函数
def introduce(name, age, city="北京"):
    print(f"我叫{name}，今年{age}岁，来自{city}")

introduce("小明", 20)
introduce("小红", 19, "上海")
introduce(age=18, name="小李", city="广州")
````

````python
# 可变参数
def sum_all(*numbers):
    return sum(numbers)

print(f"1+2+3+4+5 = {sum_all(1, 2, 3, 4, 5)}")

def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="张三", age=20, city="北京")
````

````python
# 函数作用域
global_var = "我是全局变量"

def test_scope():
    local_var = "我是局部变量"
    print(f"函数内全局变量: {global_var}")
    print(f"函数内局部变量: {local_var}")

test_scope()
print(f"函数外全局变量: {global_var}")

# 修改全局变量
counter = 0

def increment():
    global counter
    counter += 1
    print(f"当前计数: {counter}")

increment()
increment()
print(f"最终计数: {counter}")
````

## 10. 综合练习
### 学生成绩管理系统

````python
def add_student(students, name, scores):
    students[name] = scores
    print(f"✓ 已添加学生: {name}")

def calculate_average(scores):
    return sum(scores) / len(scores)

def print_student_info(students):
    print("\n" + "="*40)
    print("学生成绩单".center(40))
    print("="*40)
    
    for name, scores in students.items():
        avg = calculate_average(scores)
        print(f"姓名: {name}")
        print(f"成绩: {scores}")
        print(f"平均分: {avg:.1f}")
        print("-"*40)

def find_top_student(students):
    if not students:
        return None, 0
    
    top_student = None
    highest_avg = 0
    
    for name, scores in students.items():
        avg = calculate_average(scores)
        if avg > highest_avg:
            highest_avg = avg
            top_student = name
    
    return top_student, highest_avg

def calculate_class_average(students):
    if not students:
        return 0
    
    total_avg = 0
    for scores in students.values():
        total_avg += calculate_average(scores)
    
    return total_avg / len(students)
````

````python
# 主程序
students = {}

print("="*40)
print("学生成绩管理系统".center(40))
print("="*40)

# 添加学生
print("\n添加学生:")
add_student(students, "张三", [85, 92, 78, 90])
add_student(students, "李四", [92, 88, 95, 87])
add_student(students, "王五", [78, 85, 82, 79])
add_student(students, "赵六", [65, 72, 68, 70])

# 显示信息
print_student_info(students)

# 找出成绩最好的学生
top_student, avg_score = find_top_student(students)
print(f"\n🏆 成绩最好的学生: {top_student}，平均分: {avg_score:.1f}")

# 班级统计
class_avg = calculate_class_average(students)
print(f"📊 班级平均分: {class_avg:.1f}")

# 更多统计
print(f"\n📈 统计信息:")
print(f"  总学生数: {len(students)}")
print(f"  最高平均分: {max(calculate_average(s) for s in students.values()):.1f}")
print(f"  最低平均分: {min(calculate_average(s) for s in students.values()):.1f}")

# 找出不及格学生
failing = [(name, calculate_average(scores)) for name, scores in students.items() if calculate_average(scores) < 60]
if failing:
    print(f"\n⚠️  不及格学生:")
    for name, avg in failing:
        print(f"  - {name}: {avg:.1f}分")
else:
    print(f"\n✅ 所有学生都及格了！")

print("\n" + "="*40)
print("程序执行完毕！".center(40))
print("="*40)
````
