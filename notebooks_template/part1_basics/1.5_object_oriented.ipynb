# Part 1: 为什么需要面向对象？


```python
# 传统方式的问题
student1_name = "张三"
student1_age = 20
student1_score = 85

student2_name = "李四"
student2_age = 21
student2_score = 90

# 问题：变量分散，难以管理
print(f"{student1_name}: {student1_score}分")
print(f"{student2_name}: {student2_score}分")
```

```python
# 面向对象的方式
class Student:
    def __init__(self, name, age, score):
        self.name = name
        self.age = age
        self.score = score

    def get_grade(self):
        return "优秀" if self.score >= 90 else "良好" if self.score >= 80 else "及格"


# 创建对象
s1 = Student("张三", 20, 85)
s2 = Student("李四", 21, 90)

print(f"{s1.name}: {s1.score}分, {s1.get_grade()}")
print(f"{s2.name}: {s2.score}分, {s2.get_grade()}")
```

# Part 3: 创建第一个类


```python
class Person:
    """人类：演示基本的类结构"""

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"我是{self.name}，今年{self.age}岁"

    def celebrate_birthday(self):
        self.age += 1
        return f"{self.name}过生日了，现在{self.age}岁"


# 创建对象
person = Person("张三", 25)
print(person.introduce())
print(person.celebrate_birthday())
print(person.introduce())
```

```python
# 理解 self
class Counter:
    def __init__(self):
        self.count = 0

    def increment(self):
        self.count += 1


c1 = Counter()
c2 = Counter()

c1.increment()
c1.increment()
c2.increment()

print(f"c1.count = {c1.count}")  # 2
print(f"c2.count = {c2.count}")  # 1
# self 让每个对象维护自己的状态
```

# Part 4: 封装（Encapsulation）


```python
class BankAccount:
    """银行账户：演示封装"""

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance  # 私有属性

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            return True
        return False

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            return True
        return False

    def get_balance(self):
        return self.__balance


account = BankAccount("张三", 1000)
account.deposit(500)
account.withdraw(300)
print(f"余额: {account.get_balance()}元")

# 无法直接访问私有属性
# print(account.__balance)  # 会报错
```

```python
# 使用 @property 装饰器
class Temperature:
    def __init__(self, celsius=0):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("温度不能低于绝对零度")
        self._celsius = value

    @property
    def fahrenheit(self):
        return self._celsius * 9/5 + 32


temp = Temperature(25)
print(f"摄氏: {temp.celsius}°C, 华氏: {temp.fahrenheit}°F")

temp.celsius = 30
print(f"摄氏: {temp.celsius}°C, 华氏: {temp.fahrenheit}°F")
```

# Part 5: 继承（Inheritance）


```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def eat(self):
        return f"{self.name}正在吃东西"

    def speak(self):
        return "某种声音"


class Dog(Animal):
    def __init__(self, name, age, breed):
        super().__init__(name, age)
        self.breed = breed

    def speak(self):  # 重写方法
        return "汪汪汪"

    def fetch(self):
        return f"{self.name}正在捡球"


class Cat(Animal):
    def speak(self):
        return "喵喵喵"


# 测试继承
dog = Dog("旺财", 3, "金毛")
cat = Cat("咪咪", 2)

print(dog.eat())  # 继承自 Animal
print(f"{dog.name}说: {dog.speak()}")  # 重写的方法
print(dog.fetch())  # Dog 特有的方法

print(f"{cat.name}说: {cat.speak()}")
```

```python
# 形状类的多态
import math


class Shape:
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


shapes = [Circle(5), Rectangle(4, 6)]

for shape in shapes:
    print(f"{shape.__class__.__name__}: 面积 = {shape.area():.2f}")
```

# Part 7: 抽象（Abstraction）


```python
from abc import ABC, abstractmethod
import numpy as np


class ActivationFunction(ABC):
    """激活函数抽象基类"""

    @abstractmethod
    def forward(self, x):
        pass

    @abstractmethod
    def backward(self, x):
        pass


class ReLU(ActivationFunction):
    def forward(self, x):
        return np.maximum(0, x)

    def backward(self, x):
        return (x > 0).astype(float)


class Sigmoid(ActivationFunction):
    def forward(self, x):
        return 1 / (1 + np.exp(-x))

    def backward(self, x):
        s = self.forward(x)
        return s * (1 - s)


# 测试
x = np.array([-2, -1, 0, 1, 2])
relu = ReLU()
sigmoid = Sigmoid()

print(f"输入: {x}")
print(f"ReLU: {relu.forward(x)}")
print(f"Sigmoid: {sigmoid.forward(x)}")
```

# Part 8: 魔法方法


```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y


v1 = Vector(1, 2)
v2 = Vector(3, 4)

print(f"v1: {v1}")
print(f"v1 + v2 = {v1 + v2}")
print(f"v1 * 3 = {v1 * 3}")
print(f"v1 == v2: {v1 == v2}")
```

```python
# __call__ 方法
class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, x):
        return x * self.factor


double = Multiplier(2)
print(f"double(5) = {double(5)}")  # 像函数一样调用
```

# Part 10: 实战 - 学生管理系统


```python
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id
        self.scores = {}

    def add_score(self, subject, score):
        if 0 <= score <= 100:
            self.scores[subject] = score
            return True
        print(f"成绩{score}超出范围")
        return False

    def get_average(self):
        if not self.scores:
            return 0
        return sum(self.scores.values()) / len(self.scores)

    def __repr__(self):
        return f"Student({self.name}, 平均分={self.get_average():.2f})"


# 测试
s1 = Student("张三", "001")
s1.add_score("数学", 85)
s1.add_score("英语", 90)
s1.add_score("计算机", 92)

print(s1)
print(f"成绩: {s1.scores}")
```

```python
class Classroom:
    def __init__(self, class_name, teacher):
        self.class_name = class_name
        self.teacher = teacher
        self.students = []

    def add_student(self, student):
        self.students.append(student)

    def get_class_average(self):
        if not self.students:
            return 0
        return sum(s.get_average() for s in self.students) / len(self.students)

    def get_ranking(self):
        ranked = sorted(
            self.students, key=lambda s: s.get_average(), reverse=True)
        return [(i+1, s) for i, s in enumerate(ranked)]


# 测试
classroom = Classroom("计算机1班", "王老师")
classroom.add_student(s1)

s2 = Student("李四", "002")
s2.add_score("数学", 78)
s2.add_score("英语", 88)
classroom.add_student(s2)

print(f"班级平均分: {classroom.get_class_average():.2f}")
print("\n班级排名:")
for rank, student in classroom.get_ranking():
    print(f"第{rank}名: {student.name}, 平均分: {student.get_average():.2f}")
```

# Part 11: 继承示例 - 研究生班级


```python
class GraduateClassroom(Classroom):
    def __init__(self, class_name, teacher, research_direction):
        super().__init__(class_name, teacher)
        self.research_direction = research_direction
        self.advisors = {}

    def assign_advisor(self, student_id, advisor_name):
        self.advisors[student_id] = advisor_name
        print(f"已分配导师: {advisor_name}")


grad_class = GraduateClassroom("AI研究生班", "李教授", "深度学习")
grad_class.add_student(s1)
grad_class.assign_advisor("001", "张导师")
print(f"研究方向: {grad_class.research_direction}")
```

# Part 12: 综合练习 - 简单神经元


```python
class Neuron:
    """简单神经元：综合运用 OOP 概念"""

    def __init__(self, n_inputs):
        self.weights = np.random.randn(n_inputs)
        self.bias = 0

    def forward(self, x):
        """前向传播"""
        z = np.dot(x, self.weights) + self.bias
        return self._relu(z)

    def _relu(self, z):
        """ReLU 激活函数（私有方法）"""
        return np.maximum(0, z)

    def __repr__(self):
        return f"Neuron(inputs={len(self.weights)})"

    def __call__(self, x):
        """让神经元可以像函数一样调用"""
        return self.forward(x)


# 测试
neuron = Neuron(3)
x = np.array([1, 2, 3])
output = neuron(x)  # 使用 __call__
print(f"输入: {x}")
print(f"输出: {output:.4f}")
print(neuron)
```
